```python
from pathlib import Path
from collections import Counter
import re

root = Path("../data/")  # Assume speakers 102, 111, and 118 are removed

label_counts: Counter[str] = Counter()

for tg_path in root.rglob("*.TextGrid"):
    content = tg_path.read_text(encoding="utf-8", errors="replace")
    # Only scan the "phones" tier section
    phones_section = content.split('name = "phones"')[-1]
    labels = re.findall(r'text = "(.*?)"', phones_section)
    label_counts.update(labels)

counts = label_counts.most_common()
for label, count in counts:
    print(f"{label!r:15} {count}")
```

```python
print(len(label_counts.most_common()))
```

```python
all_files = list(root.rglob("*.TextGrid"))
files_with_empty = 0
total_empty_labels = 0

for tg_path in all_files:
    content = tg_path.read_text(encoding="utf-8", errors="replace")
    phones_section = content.split('name = "phones"')[-1]
    labels = re.findall(r'text = "(.*?)"', phones_section)
    empty_count = labels.count("")
    if empty_count > 0:
        files_with_empty += 1
        total_empty_labels += empty_count


print(f"Total files: {len(all_files)}")
print(f"Files with empty phone labels: {files_with_empty}")
print(f"Total empty phone label occurrences: {total_empty_labels}")
```

It turns out all `.TextGrid` files contain exactly one empty phone label. By quickly inspecting the files, we can see that they occur at the end of recordings.
